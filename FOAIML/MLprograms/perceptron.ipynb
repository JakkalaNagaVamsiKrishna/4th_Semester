```python
import numpy as np 

class Perceptron:
    def __init__(self, n_iters=100):
        self.n_iters = n_iters
        self.weights = None 
        self.bias = None 
    def fit(self, X, y):
        n_samples, n_features = X.shape 

        # 1. Start with weights as zeros
        self.weights = np.zeros(n_features)
        self.bias = 0

        #2. Loop until convergence or max iterations 
        for _ in range(self.n_iters):
            converged = True
            for idx, x_i in enumerate(X):
                # Calculate pre-activation: w^T * x + b 
                linear_output = np.dot(x_i, self.weights) + self.bias 

                # Predict using the classification rule 
                y_predicted = 1 if linear_output >= 0 else -1 

                # 3. If x_i is misclassified, update 
                if y_predicted != y[idx]:
                    # Update rule: w = w + y_i * x_i 
                    update = y[idx] 
                    self.weights += update*x_i 
                    self.bias += update 
                    converged = False 

            if converged:
                break

    def predict(self, X):
        linear_output = np.dot(X, self.weights) + self.bias 
        return np.where(linear_output >= 0, 1, -1)
```

```python
# Example: AND Gate
X = np.array([[0,0], [0,1], [1,0], [1,1]])
y = np.array([0, 0, 0, 1])

p = Perceptron( n_iters=10)
p.fit(X, y)

print(f"Predictions: {p.predict(X)}")
# Output should be [0, 0, 0, 1]
```
